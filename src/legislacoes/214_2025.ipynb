### LIVRO I — Do IBS e da CBS

> **Título I — Das Normas Gerais do IBS e da CBS**  
> **Capítulo I — Disposições Preliminares**  
> Arts. 1º ao 3º

---

### Art. 1º — Instituição do IBS e da CBS

> *Ficam instituídos:*
> - **I** — o **Imposto sobre Bens e Serviços (IBS)**, de competência compartilhada entre Estados, Municípios e Distrito Federal (CF, art. 156-A);
> - **II** — a **Contribuição Social sobre Bens e Serviços (CBS)**, de competência da União (CF, art. 195, V).

#### 💬 Meu comentário

O Art. 1º é o marco zero da reforma: dois tributos novos nascem juntos, mas com competências distintas.
O **IBS** substitui o ICMS (estadual) e o ISS (municipal) — reunindo numa só incidência o que hoje é fragmentado em 27 sistemas estaduais e mais de 5.000 municipais.
A **CBS** substitui o PIS e a Cofins federais.

A separação de competência (subnacional × federal) foi o principal nó político da reforma;
a solução adotada foi um tributo *dual* com legislação unificada mas arrecadação compartilhada.
Isso explica por que os dois tributos aparecem sempre em par ao longo de toda a lei.

### Art. 2º — Princípio da Neutralidade

> *O IBS e a CBS são informados pelo princípio da neutralidade, segundo o qual esses tributos devem evitar distorcer as decisões de consumo e de organização da atividade econômica, observadas as exceções previstas na CF e nesta Lei Complementar.*

#### 💬 Meu comentário

A **neutralidade** é o princípio-chave de um IVA moderno e aparece aqui como norma positiva, não apenas como objetivo de política.

Na prática, neutralidade significa:

- **Não-cumulatividade plena** — o imposto pago em cada etapa é creditado, de modo que a carga total recai apenas sobre o consumidor final, independentemente do tamanho da cadeia produtiva.
- **Tratamento uniforme entre setores** — a mesma alíquota incide sobre bens e serviços equivalentes, eliminando a vantagem fiscal de escolher uma forma de organização em vez de outra.

A ressalva *"observadas as exceções"* é importante: a própria CF autoriza regimes diferenciados (saúde, educação, agronegócio, Zona Franca de Manaus etc.), que serão tratados em artigos posteriores. A neutralidade é, portanto, a regra geral — as exceções precisam de fundamento constitucional expresso.

### Art. 3º — Definições Fundamentais

> Este artigo estabelece o vocabulário jurídico da lei, definindo os conceitos de **operação**, **fornecimento**, **fornecedor**, **adquirente** e **destinatário**.

#### Mapa dos conceitos

| Conceito | Definição resumida |
|---|---|
| **Operação com bens** | Qualquer ato envolvendo bens móveis, imóveis, materiais, imateriais ou direitos |
| **Operação com serviços** | Tudo que *não* for operação com bens (conceito residual) |
| **Fornecimento** | Entrega, transferência, prestação ou disponibilização — o fato gerador em sentido amplo |
| **Fornecedor** | Quem realiza o fornecimento (PF, PJ, residente no país ou exterior) |
| **Adquirente** | Quem tem obrigação de pagar pelo fornecimento |
| **Destinatário** | A quem o bem/serviço é efetivamente entregue (pode ≠ adquirente) |

#### 💬 Meu comentário

**Por que definir tudo isso logo no Art. 3º?**
O IBS/CBS incide sobre o *fornecimento* — conceito propositalmente amplo para evitar as brechas do sistema atual.
No ICMS, por exemplo, a discussão sobre "serviço vs. mercadoria" gerou décadas de litígios (software, streaming, leasing etc.).
A LC 214 tenta fechar esse espaço ao dizer: se não é operação com bem, é serviço — ponto final.

**Destaque: § 1º** — energias com valor econômico (eletricidade, gás, vapor) são equiparadas a bens materiais, consolidando entendimento já pacificado pelo STJ.

**Destaque: § 2º** — fundos de investimento e consórcios entram como "fornecedores", o que tem impacto relevante no mercado financeiro e imobiliário.

**Destaque: § 3º** *(incluído pela LC 227/2026)* — locação, arrendamento e cessão temporária de bens estão expressamente no campo de incidência.
Isso resolve a histórica controvérsia sobre o ISS na locação de bens móveis.

## Capítulo II — Do IBS e da CBS sobre Operações com Bens e Serviços

### Seção I — Das Hipóteses de Incidência

### Art. 4º — Fato Gerador: Operações Onerosas

> *O IBS e a CBS incidem sobre operações onerosas com bens ou com serviços.*

#### 💬

O **fato gerador** é a operação onerosa — conceito propositalmente amplo que engloba oito modalidades expressas no § 2º:

| # | Modalidade |
|---|---|
| I | Compra e venda, troca, permuta, dação em pagamento |
| II | Locação |
| III | Licenciamento, concessão, cessão |
| IV | Mútuo oneroso |
| V | Doação com contraprestação em benefício do doador |
| VI | Instituição onerosa de direitos reais |
| VII | Arrendamento (inclusive mercantil) |
| VIII | Prestação de serviços |

**O § 3º é especialmente relevante:** a lei declara irrelevantes o título jurídico da posse, a validade jurídica do negócio, a obtenção de lucro e o cumprimento de formalidades legais.
Traduzindo: **a incidência é sobre a substância econômica, não sobre a forma jurídica**.
Isso fecha uma das principais portas de planejamento tributário agressivo do sistema atual.

**§ 4º** — A incidência alcança operações com **ativo não circulante** (máquinas, imóveis, veículos) e atividades **não habituais**.
No ICMS, a venda ocasional de ativo imobilizado gerava discussões intermináveis sobre habitualidade; aqui o texto é expresso.

**§ 5º** — A incidência do IBS/CBS **não altera a base de cálculo do ITCD e do ITBI**.
Isso evita dupla tributação em transmissões patrimoniais — o IBS/CBS entra pela contraprestação (serviço de intermediação, por ex.), não pelo valor do bem transmitido.

### Art. 5º — Incidência em Operações Não Onerosas ou Abaixo do Mercado

> *O IBS e a CBS também incidem sobre fornecimentos não onerosos ou a valor inferior ao de mercado.*

#### 💬

Este artigo implementa o mecanismo anti-elisão para **operações entre partes relacionadas e benefícios a pessoas físicas vinculadas** ao contribuinte.

A lógica é: se houve crédito na entrada, deve haver débito na saída — mesmo que a saída seja gratuita ou subfaturada.
Sem essa regra, seria simples transformar consumo pessoal em despesa empresarial com crédito de IBS/CBS.

**Quem é alcançado pelo inc. I (alíneas "a" e "b"):**

- O próprio contribuinte pessoa física
- Sócios, acionistas, administradores, membros de conselho
- Empregados
- Cônjuges/companheiros e parentes até **3º grau** de todos os acima

**Inc. II — Brindes e bonificações:**
A regra geral é tributar, *salvo* quando a bonificação conste do documento fiscal e não dependa de evento posterior (ex.: bonificação em nota de venda).

**Inc. III — Devolução de capital / dividendos in natura:**
Se o bem distribuído gerou crédito quando entrou na empresa, sua distribuição ao sócio é fato gerador.
Evita que a empresa compre bens, tome crédito e depois os distribua sem tributação de saída.

**§ 9º** — A base de cálculo nesses casos é o **valor de mercado**, não o valor cobrado (que pode ser zero).

**Partes relacionadas (§§ 2º a 6º):**
O conceito é amplo e detalhado — controle direto/indireto, participação ≥ 20% do capital, sócios comuns, parentesco com diretores.
Vale atenção especial para **fundos de investimento** e **grupos multinacionais**, que entram pela definição do § 3º, III.

### Art. 6º — Não Incidência

> *O IBS e a CBS não incidem sobre doze hipóteses expressamente listadas.*

#### 💬 

O Art. 6º delimita o campo negativo da incidência. Agrupo as hipóteses por lógica:

**Relações internas da empresa (incs. I e II)**
- Salários e remuneração de administradores → fora do campo (não é fornecimento)
- Transferência entre estabelecimentos do **mesmo contribuinte** → não é operação de mercado

**Operações societárias e financeiras (incs. III a VII)**
- Alienação de participação societária, fusão, cisão, incorporação, integralização de capital → fora do campo
- Rendimentos financeiros → fora, *exceto* quando o contribuinte está no regime específico de serviços financeiros
- Dividendos e JCP → fora do campo

> ⚠️ **Ressalva do § 1º:** se o conjunto de atos, na *essência*, constituir uma operação onerosa disfarçada (ex.: reestruturação societária que encobre uma venda), o IBS/CBS incide. Substância prevalece sobre forma — coerente com o § 3º do Art. 4º.

**Setor público e terceiro setor (incs. VIII e IX)**
- Doações sem contraprestação → fora do campo
- Transferências de recursos públicos para OSCs (convênios, termos de fomento, contratos de gestão) → fora do campo

**Cooperativismo (incs. X e XI)**
- Destinação para fundos obrigatórios (FATES, reserva de assistência) → fora do campo
- Repasse aos associados e distribuição de sobras → fora do campo

**Contribuições associativas (inc. XII — LC 227/2026)**
- Mensalidades e anuidades de associações civis sem fins econômicos → fora do campo, desde que preenchidos os requisitos do CTN (art. 14)

> 💡 **Ponto de atenção:** o § 2º prevê que, se uma doação (inc. VIII) envolver bens que geraram crédito, o doador deverá ou tributar pelo valor de mercado *ou* anular os créditos. Não existe saída gratuita de crédito.
